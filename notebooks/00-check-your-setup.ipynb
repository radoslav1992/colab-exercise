## What can this browser do?

`lab` is the small Python package of this course. `lab.device()` asks the browser what it offers.

- `webgpu: True` means models can run on your graphics card. This needs a recent Chrome or Edge (or another browser with WebGPU).
- `webgpu: False` means models fall back to the processor (`wasm`). Everything still works, only slower.

```python
import lab

info = lab.device()
for key, value in info.items():
    print(f"{key:>22}: {value}")
```

## A tiny matrix example

A matrix times a vector, and solving a linear system $A x = b$:

$$A = \begin{pmatrix} 2 & 1 \\ 1 & 3 \end{pmatrix}, \qquad b = \begin{pmatrix} 3 \\ 5 \end{pmatrix}$$

By hand: from the first row $x_2 = 3 - 2 x_1$; put into the second row, $x_1 + 9 - 6 x_1 = 5$, so $x_1 = 0.8$ and $x_2 = 1.4$.

```python
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
b = np.array([3.0, 5.0])

x = np.linalg.solve(A, b)
print("x =", x)
print("A @ x =", A @ x)

lab.table([{"": "A", "shape": A.shape, "dtype": A.dtype}, {"": "b", "shape": b.shape, "dtype": b.dtype}, {"": "x", "shape": x.shape, "dtype": x.dtype}])
```

## A check cell

Cells with a coloured bar on the left are **check cells**: they test your work with `assert` and show *Check passed* or *Check failed*. The line at the top of the page counts them. You will meet them in every assignment.

```python
assert np.allclose(x, [0.8, 1.4]), "x should be (0.8, 1.4)"
assert np.allclose(A @ x, b), "A @ x should give b back"
print("Your setup works.")
```

## A picture

`lab.show` displays an array as an image: here a 64 × 64 grey ramp with a bright circle.

```python
yy, xx = np.mgrid[0:64, 0:64]
ramp = xx / 63.0
circle = (xx - 32) ** 2 + (yy - 32) ** 2 < 15 ** 2
image = np.where(circle, 1.0, 0.6 * ramp)
lab.show(np.kron(image, np.ones((3, 3))))   # every pixel enlarged 3 × 3
```
